## 2.6 `scipy.linalg`: Álgebra Linear Avançada

```python
from scipy.linalg import lu
from scipy import stats
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

# Matriz de exemplo


```python
A = np.array([[3, 2, 1], [1, 1, 1], [0, 1, 2]])
```

# Decomposição LU


```python
P, L, U = lu(A)
print("Matriz L:")
print(L)
print("Matriz U:")
print(U)
```
