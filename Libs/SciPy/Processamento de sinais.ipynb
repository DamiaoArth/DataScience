## 2.7 `scipy.signal`: Processamento de Sinais


```python
from scipy.signal import find_peaks
from scipy import stats
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

# Dados de exemplo


```python
x = np.linspace(0, 6*np.pi, 100)
y = np.sin(x)
```

# Encontrando picos

```python
peaks, _ = find_peaks(y, height=0)
```

# Visualização

```python
plt.plot(x, y)
plt.plot(x[peaks], y[peaks], "x")
plt.title("Picos em um Sinal")
plt.show()
```
