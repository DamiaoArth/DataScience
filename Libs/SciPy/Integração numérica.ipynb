## 2.5 `scipy.integrate`: Integração Numérica

```python
from scipy.integrate import quad
from scipy import stats
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

# Função para integrar


```python
f = lambda x: x**2
```

# Integração


```python
resultado, erro = quad(f, 0, 1)
print(f"Resultado da integração: {resultado:.3f}")
```
