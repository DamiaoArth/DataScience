## Problema: Análise de Dados e Otimização Multimodal

Neste exemplo, resolveremos um problema que combina estatísticas, interpolação, e otimização usando SciPy:

```python
from scipy.optimize import differential_evolution
from scipy import stats
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

# Função multimodal


```python
f = lambda x: np.sin(5 * x) * (1 - np.tanh(x**2))
```

# Visualização da função


```python
x = np.linspace(-2, 2, 500)
y = f(x)
plt.plot(x, y)
plt.title("Função Multimodal")
plt.show()
```

# Otimização Mínima global


```python
bounds = [(-2, 2)]
result = differential_evolution(f, bounds)
print(f"Mínimo global: {result.fun:.3f} em x = {result.x[0]:.3f}")
```

# Visualizando o resultado


```python
plt.plot(x, y)
plt.scatter(result.x, result.fun, color="red", label="Mínimo Global")
plt.legend()
plt.title("Otimização Global")
plt.show()
```

# Otimização Máxima global


```python
bounds = [(-2, 2)]
result = differential_evolution(lambda x: -f(x), bounds)
max_value = -result.fun
print(f"Máximo global: {max_value:.3f} em x = {result.x[0]:.3f}")
```

# Visualizando o resultado


```python
plt.plot(x, y)
plt.scatter(result.x, max_value, color="blue", label="Máximo Global")
plt.legend()
plt.title("Otimização Global - Máximo")
plt.show()
```
