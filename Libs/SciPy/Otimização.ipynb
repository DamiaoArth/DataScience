## 2.2 `scipy.optimize`: Otimização


```python
from scipy.optimize import minimize
from scipy import stats
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

# Função objetivo

```python
f = lambda x: (x - 3)**2 + 5
```

# Otimização

```python
result = minimize(f, x0=0)
print(f"Valor mínimo: {result.fun:.2f} em x = {result.x[0]:.2f}")
```

# Visualizando a função

```python
x = np.linspace(-2, 8, 100)
y = f(x)
plt.plot(x, y, label="Função Objetivo")
plt.scatter(result.x, result.fun, color="red", label="Mínimo")
plt.legend()
plt.title("Otimização de Função")
plt.show()
```
