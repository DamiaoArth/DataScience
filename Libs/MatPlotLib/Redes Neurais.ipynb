# Introdução a Redes Neurais com scikit-learn

Neste notebook, vamos explorar conceitos básicos de redes neurais para tarefas de classificação. Utilizaremos o `MLPClassifier` da biblioteca scikit-learn para implementar exemplos simples.

Tópicos abordados:
1. **Introdução ao Perceptron** usando o `MLPClassifier`.
2. Exemplos básicos de classificação com redes neurais simples.

# Importando bibliotecas

```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import classification_report, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
```

# 1. Criando um Dataset Sintético

## Gerando dados para classificação binária

```python
data, labels = make_classification(n_samples=1000, n_features=10, n_informative=5, n_redundant=2, n_classes=2, random_state=42)
```

# Dividindo os dados em treino e teste

```python
X_train, X_test, y_train, y_test = train_test_split(data, labels, test_size=0.3, random_state=42, stratify=labels)
```

```python
print("Dados de treino e teste preparados!")
```

```python
# 2. Introdução ao Perceptron com `MLPClassifier`

## Definindo e treinando o modelo
```

```python
mlp = MLPClassifier(hidden_layer_sizes=(10,), max_iter=1000, random_state=42)
mlp.fit(X_train, y_train)
```

# Avaliando o modelo

```python
train_score = mlp.score(X_train, y_train)
test_score = mlp.score(X_test, y_test)
```

```python
print(f"Acurácia no conjunto de treino: {train_score:.2f}")
print(f"Acurácia no conjunto de teste: {test_score:.2f}")
```

# Relatório de classificação

```python
y_pred = mlp.predict(X_test)
```

```python
print("\nRelatório de classificação:")
print(classification_report(y_test, y_pred))
```

# Matriz de confusão

```python
ConfusionMatrixDisplay.from_estimator(mlp, X_test, y_test)
plt.show()
```

# 3. Modificando a Arquitetura da Rede Neural

## Adicionando mais camadas ocultas e neurônios

```python
mlp_complex = MLPClassifier(hidden_layer_sizes=(50, 20), max_iter=1000, random_state=42)
mlp_complex.fit(X_train, y_train)
```

# Avaliando o modelo com arquitetura modificada

```python
train_score_complex = mlp_complex.score(X_train, y_train)
test_score_complex = mlp_complex.score(X_test, y_test)
```

```python
print(f"Acurácia (rede complexa) no conjunto de treino: {train_score_complex:.2f}")
print(f"Acurácia (rede complexa) no conjunto de teste: {test_score_complex:.2f}")
```

# Relatório de classificação (rede complexa)

```python
y_pred_complex = mlp_complex.predict(X_test)
```

```python
print("\nRelatório de classificação (rede complexa):")
print(classification_report(y_test, y_pred_complex))
```

# Matriz de confusão (rede complexa)

```python
ConfusionMatrixDisplay.from_estimator(mlp_complex, X_test, y_test)
plt.show()
```
