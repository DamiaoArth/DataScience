# Pré-processamento de Dados com scikit-learn

Neste notebook, exploraremos técnicas fundamentais para preparar dados para modelagem de machine learning usando **scikit-learn**.

Técnicas abordadas:
1. **Normalização de Dados**: Ajusta os valores para uma escala padrão.
2. **Codificação de Variáveis Categóricas**: Transforma texto em números.
3. **Tratamento de Valores Ausentes**: Substitui ou remove dados faltantes.


# Importando as bibliotecas necessárias

```python
from sklearn.preprocessing import StandardScaler, LabelEncoder, OneHotEncoder
from sklearn.impute import SimpleImputer
import numpy as np
import pandas as pd
```

# Criando um conjunto de dados exemplo

```python
dados_exemplo = pd.DataFrame({
    'Idade': [25, np.nan, 35, 40, 29],
    'Salário': [50000, 60000, np.nan, 80000, 50000],
    'Cidade': ['São Paulo', 'Rio de Janeiro', 'Belo Horizonte', 'São Paulo', 'Rio de Janeiro'],
    'Gênero': ['Masculino', 'Feminino', 'Feminino', np.nan, 'Masculino']
})
```

```python
print("Conjunto de dados inicial:")
print(dados_exemplo)
```

## 1. Tratamento de Valores Ausentes
Utilizamos `SimpleImputer` para preencher os valores ausentes com estratégias como média, mediana ou moda.


# Tratamento de valores ausentes

```python
imputer_idade = SimpleImputer(strategy='mean')
dados_exemplo['Idade'] = imputer_idade.fit_transform(dados_exemplo[['Idade']])
```

```python
imputer_salario = SimpleImputer(strategy='median')
dados_exemplo['Salário'] = imputer_salario.fit_transform(dados_exemplo[['Salário']])
```

```python
print("\nDados após tratamento de valores ausentes:")
print(dados_exemplo)
```

## 2. Normalização dos Dados Numéricos
A normalização padroniza os valores em uma distribuição com média 0 e desvio padrão 1 usando `StandardScaler`.


# Normalização de Idade e Salário

```python
scaler = StandardScaler()
dados_numericos = scaler.fit_transform(dados_exemplo[['Idade', 'Salário']])
```

```python
dados_exemplo[['Idade Normalizada', 'Salário Normalizado']] = dados_numericos
```

```python
print("\nDados após normalização:")
print(dados_exemplo)
```

## 3. Codificação de Variáveis Categóricas
### 3.1 Codificação com LabelEncoder
Transforma valores categóricos em valores inteiros.


# Codificação de Gênero com LabelEncoder

```python
le = LabelEncoder()
dados_exemplo['Gênero Codificado'] = le.fit_transform(dados_exemplo['Gênero'].fillna('Desconhecido'))
```

```python
print("\nDados após codificação de Gênero com LabelEncoder:")
print(dados_exemplo)
```

### 3.2 Codificação One-Hot
`OneHotEncoder` cria uma coluna binária para cada valor único.


# Codificação de Cidade com OneHotEncoder

```python
ohe = OneHotEncoder(sparse_output=False)
cidades_codificadas = ohe.fit_transform(dados_exemplo[['Cidade']])
```

```python
nomes_colunas = ohe.get_feature_names_out(['Cidade'])
dados_cidades = pd.DataFrame(cidades_codificadas, columns=nomes_colunas)
```

# Combinando com o DataFrame original

```python
dados_final = pd.concat([dados_exemplo, dados_cidades], axis=1)
```

```python
print("\nDados finais com codificação One-Hot:")
print(dados_final)
```
