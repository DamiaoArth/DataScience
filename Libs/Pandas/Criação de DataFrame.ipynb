## Criação de Dataframes

## Instalando e importando Dependêcias

```python
%pip install pandas 
import pandas as pd
```

## Criando um DataFrame a partir de uma lista

```python
dados = [[1, 'Arthur', 'Curitiba'], [2, 'Bianca', 'Curitiba'], [3, 'Carlos', 'Curitiba']]

df = pd.DataFrame(dados, columns=['ID', 'Nome', 'Cidade'])

df.head() #para o início
#df.tail() para o final
```

## Criando um DataFrame a partir de um dicionário

```python
dados = {'ID': [1, 2, 3], 'Nome': ['Arthur', 'Bianca', 'Carlos'], 'Cidade': ['Curitiba', 'Curitiba', 'Curitiba']}

df = pd.DataFrame(dados)

df.head() #para o início
#df.tail() para o final
```

## Criando um DataFrame a partir de uma lista de dicionários

```python
dados = [{'ID': 1, 'Nome': 'Arthur', 'Cidade': 'Curitiba'}, {'ID': 2, 'Nome': 'Bianca', 'Cidade': 'Curitiba'}, {'ID': 3, 'Nome': 'Carlos', 'Cidade': 'Curitiba'}]

df = pd.DataFrame(dados)

df.head() #para o início
#df.tail() para o final
```

## Salvar **DF** em um arquivo CSV 

```python
df.to_csv('dados.csv', index=False, encoding='latin1')
```

## Salvar **DF** em um arquivo Excel

```python
df.to_excel('dados.xlsx', index=False, encoding='latin1')
```
