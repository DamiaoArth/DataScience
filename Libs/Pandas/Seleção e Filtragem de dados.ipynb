## Definição de um **DF** e leitura dos arquivos **HTML**

```python
df = pd.read_html('https://gist.github.com/DamiaoArth/ace0ed235f37244ef5d883e0b6354744')[0]
```

```python
df
```

```python
coluna = df['Name']

colunas= df[['Name', 'Type1', 'Attack']]

linhas = df[df['Attack'] > 100]

# Filtrar HP
HP = df.loc[df['HP'] > 150, ['Name', 'Type1', 'Type2', 'HP']].sort_values(by='HP', ascending=False)

# Filtrar Attack
Attack = df.loc[df['Attack'] > 150, ['Name', 'Type1', 'Type2', 'Attack']].sort_values(by='Attack', ascending=False)

# Filtrar Defense
Defense = df.loc[df['Defense'] > 150, ['Name', 'Type1', 'Type2', 'Defense']].sort_values(by='Defense', ascending=False)

# Filtrar Special Attack (SP_Attack)
SP_Attack = df.loc[df['SP_Attack'] > 150, ['Name', 'Type1', 'Type2', 'SP_Attack']].sort_values(by='SP_Attack', ascending=False)

# Filtrar Special Defense (SP_Defense)
SP_Defense = df.loc[df['SP_Defense'] > 150, ['Name', 'Type1', 'Type2', 'SP_Defense']].sort_values(by='SP_Defense', ascending=False)

# Filtrar Speed
Speed = df.loc[df['Speed'] > 150, ['Name', 'Type1', 'Type2', 'Speed']].sort_values(by='Speed', ascending=False)
```
