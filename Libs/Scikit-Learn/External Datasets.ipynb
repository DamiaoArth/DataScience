# Trabalhando com Conjuntos de Dados Externos

Neste notebook, exploraremos como carregar e trabalhar com diferentes tipos de datasets externos, incluindo arquivos locais e fontes online. Os tópicos abordados incluem:

1. Carregando datasets de arquivos **CSV** e **Excel**.
2. Conectando-se a **APIs** para carregar dados dinâmicos.
3. Utilizando datasets públicos de plataformas como o Kaggle.

# Importando bibliotecas

```python
import pandas as pd
import requests
from io import StringIO
```

# 1. Carregando Arquivos Locais (CSV e Excel)

## Carregando um arquivo CSV

```python
csv_file_path = 'data/exemplo.csv'  # Certifique-se de que o arquivo exista no caminho especificado
```

```python
try:
    df_csv = pd.read_csv(csv_file_path)
    print("\nDados do CSV:")
    print(df_csv.head())
```

```python
except FileNotFoundError:
    print("Arquivo CSV não encontrado. Certifique-se de fornecer um caminho válido.")
```

## Carregando um arquivo Excel

```python
excel_file_path = 'data/exemplo.xlsx'  # Certifique-se de que o arquivo exista no caminho especificado
```

```python
try:
    df_excel = pd.read_excel(excel_file_path)
    print("\nDados do Excel:")
    print(df_excel.head())
```

```python
except FileNotFoundError:
    print("Arquivo Excel não encontrado. Certifique-se de fornecer um caminho válido.")
```

# 2. Carregando Dados de uma API

## Exemplo com uma API pública

```python
url = 'https://api.exemplo.com/dados'  # Substitua por uma API válida
```

```python
try:
    response = requests.get(url)
    if response.status_code == 200:
        data = StringIO(response.text)
        df_api = pd.read_csv(data)
        print("\nDados carregados da API:")
        print(df_api.head())
    else:
        print(f"Erro ao acessar a API: {response.status_code}")
```

```python
except Exception as e:
    print(f"Erro ao conectar-se à API: {e}")
```

# 3. Trabalhando com Datasets Públicos (Kaggle)

```python
kaggle_file_path = 'data/kaggle_dataset.csv'  # Altere para o caminho do dataset baixado do Kaggle
```

```python
try:
    df_kaggle = pd.read_csv(kaggle_file_path)
    print("\nDados do Kaggle:")
    print(df_kaggle.head())
```

```python
except FileNotFoundError:
    print("Arquivo Kaggle não encontrado. Certifique-se de que o dataset foi baixado corretamente.")
```
