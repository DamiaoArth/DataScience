# Regressão Linear e Logística

## Regressão Linear
A regressão linear modela a relação entre uma variável dependente contínua e uma ou mais independentes.

### Fórmula:
\[ y = \beta_0 + \beta_1x + \epsilon \]

## Regressão Logística
A regressão logística é usada para problemas de classificação binária.

### Fórmula:
\[ P(y=1) = \frac{1}{1 + e^{-(\beta_0 + \beta_1x)}} \]

## Diferenças
- Linear: Previsão contínua.
- Logística: Previsão categórica.

```python
boston = load_boston()
X = boston.data
y = boston.target
```

# Dividir em treino e teste

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

# Treinar o modelo

```python
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

# Previsões

```python
y_pred = regressor.predict(X_test)
```

# Calcular RMSE

```python
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
print(f"RMSE: {rmse}")
```

## Visualização:
Você pode visualizar a linha de regressão sobre os dados de teste.

```python
plt.scatter(y_test, y_pred)
plt.xlabel("Valores reais")
plt.ylabel("Valores previstos")
plt.title("Regressão Linear: Valores Reais vs Previstos")
plt.show()
```

## Regressão Logística
### Objetivo
Usar a regressão logística para classificar dados em duas categorias.

### Passos:
Carregar os dados: Use um dataset binário, como o load_breast_cancer do sklearn.datasets.
Dividir os dados: Separe em treino e teste.
Ajustar o modelo: Use LogisticRegression do sklearn.linear_model.
### Avaliar: Calcule a AUC (Área sob a Curva) para medir a performance do modelo.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_curve
```

### Carregar dados

```python
cancer = load_breast_cancer()
X = cancer.data
y = cancer.target
```

### Dividir em treino e teste

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

### Treinar o modelo

```python
logreg = LogisticRegression(max_iter=10000)
logreg.fit(X_train, y_train)
```

### Previsões de probabilidade

```python
y_prob = logreg.predict_proba(X_test)[:, 1]
```

### Calcular AUC

```python
auc = roc_auc_score(y_test, y_prob)
print(f"AUC: {auc}")
```

### Visualização:
Você pode plotar a curva ROC para visualização da performance.

```python
fpr, tpr, _ = roc_curve(y_test, y_prob)

plt.figure()
plt.plot(fpr, tpr, color='darkorange', lw=2, label='Curva ROC (area = %0.2f)' % auc)
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('Taxa de Falsos Positivos')
plt.ylabel('Taxa de Verdadeiros Positivos')
plt.title('Curva ROC')
plt.legend(loc="lower right")
plt.show()
```
