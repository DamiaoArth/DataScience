# Estatística

## Introdução
A estatística é uma ciência voltada para o estudo de dados, oferecendo ferramentas para coleta, análise, interpretação e apresentação dos mesmos.

### Tipos de Estatística
1. **Descritiva**: Resumo e descrição dos dados (ex.: média, mediana, moda, gráficos).
2. **Inferencial**: Uso de uma amostra para fazer generalizações sobre uma população.

### Medidas Descritivas
- **Tendência Central**: Média, Mediana, Moda.
- **Dispersão**: Variância, Desvio Padrão, Amplitude.
- **Distribuição de Frequências**: Histogramas, tabelas de frequência.

### Probabilidade
A probabilidade é a base da estatística inferencial, usada para modelar incertezas e prever eventos.

### Visualização de Dados
- Histogramas
- Boxplots
- Diagramas de dispersão


### Importando bibliotecas necessárias

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
```

### Criando um conjunto de dados de exemplo

```python
np.random.seed(42)
data = np.random.normal(loc=50, scale=10, size=1000)  # 1000 valores com média 50 e desvio padrão 10
```


### Cálculo de medidas estatísticas


```python
media = np.mean(data)
mediana = np.median(data)
moda = stats.mode(data).mode[0]
variancia = np.var(data, ddof=1)  # Variância amostral
desvio_padrao = np.std(data, ddof=1)  # Desvio padrão amostral
```


### Exibindo resultados

```python

resultados_estatisticos = {
    "Média": media,
    "Mediana": mediana,
    "Moda": moda,
    "Variância": variancia,
    "Desvio Padrão": desvio_padrao,
}
resultados_estatisticos
```

### Gerando o histograma

```python
plt.figure(figsize=(10, 6))
sns.histplot(data, bins=30, kde=True, color='blue', alpha=0.7)
plt.axvline(media, color='red', linestyle='--', label='Média')
plt.axvline(mediana, color='green', linestyle='-', label='Mediana')
plt.axvline(moda, color='purple', linestyle='-', label='Moda')
plt.title("Histograma dos Dados")
plt.xlabel("Valores")
plt.ylabel("Frequência")
plt.legend()
plt.show()
```

### Gerando o boxplot

```python
plt.figure(figsize=(8, 4))
sns.boxplot(x=data, color='cyan')
plt.title("Boxplot dos Dados")
plt.xlabel("Valores")
plt.show()
```
