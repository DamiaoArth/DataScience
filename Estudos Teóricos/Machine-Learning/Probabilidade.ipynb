# Probabilidade

## Introdução
A probabilidade mede a chance de ocorrência de um evento. É representada como um valor entre 0 e 1.

## Regras de Probabilidade
- Soma: \( P(A \cup B) = P(A) + P(B) - P(A \cap B) \)
- Multiplicação: \( P(A \cap B) = P(A) \cdot P(B|A) \)

## Variáveis Aleatórias
- **Discreta**: Valores finitos (ex.: número de vezes que um dado é lançado).
- **Contínua**: Intervalos contínuos (ex.: tempo).

## Distribuições de Probabilidade
- **Binomial**: Distribuição de sucesso/falha.
- **Normal**: Distribuição em forma de sino.



# Exercícios: Probabilidade

### Importando bibliotecas necessárias


```python
from scipy.stats import binom, norm
```


## 1. Geração de distribuições usando numpy e scipy

### Distribuição Binomial: 10 lançamentos de uma moeda justa


```python
n = 10  # número de experimentos
p = 0.5  # probabilidade de sucesso
x = np.arange(0, n + 1)
pmf_binomial = binom.pmf(x, n, p)
```


### Distribuição Normal: média 0, desvio padrão 1

```python

x_norm = np.linspace(-4, 4, 1000)
pdf_normal = norm.pdf(x_norm, loc=0, scale=1)
```


### Plotando distribuições

```python

plt.figure(figsize=(14, 6))
plt.subplot(1, 2, 1)
plt.bar(x, pmf_binomial, color='blue', alpha=0.7, label='Binomial (n=10, p=0.5)')
plt.title("Distribuição Binomial")
plt.xlabel("Número de Sucessos")
plt.ylabel("Probabilidade")
plt.legend()
```

```python


plt.subplot(1, 2, 2)
plt.plot(x_norm, pdf_normal, color='red', label='Normal (µ=0, σ=1)')
plt.title("Distribuição Normal")
plt.xlabel("Valores")
plt.ylabel("Densidade de Probabilidade")
plt.legend()
```

```python


plt.tight_layout()
plt.show()
```



## 2. Simulação de Experimentos Probabilísticos

### Simulando 1000 lançamentos de moeda (cara=1, coroa=0)

```python

moeda = np.random.choice([0, 1], size=1000, p=[0.5, 0.5])
caras = np.sum(moeda == 1)
coroas = np.sum(moeda == 0)
```


# Exibindo os resultados

```python

simulacao_resultados = {
    "Total de Lançamentos": len(moeda),
    "Caras": caras,
    "Coroas": coroas,
}
simulacao_resultados
```
