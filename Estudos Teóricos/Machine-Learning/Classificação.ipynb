# Classificação

## Introdução
A classificação é uma técnica de aprendizado supervisionado usada para categorizar dados.

## Algoritmos Comuns
1. **K-Nearest Neighbors (KNN)**: Baseado na proximidade dos dados.
2. **SVM (Support Vector Machines)**: Maximiza a margem entre classes.
3. **Árvores de Decisão**: Divide os dados em critérios baseados em condições.

## Métricas de Avaliação
- Acurácia
- Precisão
- Recall
- F1-Score



## Exercício: Classificação com KNN

### Importando bibliotecas necessárias

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix
```

### Gerando um dataset fictício para classificação

```python
X, y = make_classification(n_samples=500, n_features=4, n_classes=2, random_state=42)
```

### Dividindo os dados em treino e teste

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

### Criando e treinando o modelo KNN

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
```

### Fazendo previsões

```python
y_pred = knn.predict(X_test)
```

### Avaliação do modelo

```python
conf_matrix = confusion_matrix(y_test, y_pred)
class_report = classification_report(y_test, y_pred)
```

### Exibindo resultados

```python
conf_matrix, class_report
```
